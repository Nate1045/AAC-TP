## 4.6 Recursion

A **recursive** function calls itself. It always needs a **base case**, otherwise it never stops.

````python
def factorial(n):
    if n <= 1:          # base case
        return 1
    return n * factorial(n - 1)

print([factorial(k) for k in range(8)])
````

# Part 5 — Measuring an algorithm (Lab 1 prep)

In **Lab 1**, for each program, you will:

1. **count** how many times the elementary operation runs;
2. **measure** the running time;
3. **plot** both quantities against $n$, on a **logarithmic scale**.

We practise here on a simple example: **summing the elements of a list**.

## 5.1 Counting operations

````python
def total(T):
    counter = 0
    s = 0
    for e in T:
        s += e
        counter += 1          # elementary operation: one addition
    return s, counter

print(total([1, 2, 3, 4, 5]))
````

## 5.2 Timing

We use `time.perf_counter()`, the most precise clock available.
A single measurement is **noisy** (other programs, caches…), so we repeat **at least 3 times** and keep the **average**.

````python
import time

def stopwatch(f, *args, repetitions=3):
    # Average time (in seconds) of f(*args) over several repetitions.
    elapsed = 0.0
    for _ in range(repetitions):
        start = time.perf_counter()
        f(*args)
        elapsed += time.perf_counter() - start
    return elapsed / repetitions

print(f"{stopwatch(total, list(range(100_000))):.6f} s")
````

## 5.3 Filling the results table

````python
sizes, operations, times = [], [], []

print(f"{'n':>10} | {'operations':>12} | {'time (s)':>10}")
print("-" * 38)
for n in [10, 100, 1_000, 10_000, 100_000, 1_000_000]:
    T = list(range(n))
    _, ops = total(T)
    t = stopwatch(total, T)
    sizes.append(n); operations.append(ops); times.append(t)
    print(f"{n:>10,} | {ops:>12,} | {t:>10.6f}")
````

## 5.4 Plotting on a log-log scale

Why **log**? $n$ ranges from $10$ to $10^6$: on a normal scale, all the small points get squashed against the axis.
On a **log-log** scale, growth proportional to $n^k$ becomes a **straight line with slope $k$**.

````python
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4))

ax1.loglog(sizes, operations, "o-")
ax1.set(title="Number of operations", xlabel="n (input size)", ylabel="operations")
ax1.grid(True, which="both", alpha=0.3)

ax2.loglog(sizes, times, "o-", color="tab:red")
ax2.set(title="Running time", xlabel="n (input size)", ylabel="time (s)")
ax2.grid(True, which="both", alpha=0.3)

plt.tight_layout()
plt.show()
````

> 🧠 **Question**: what do you notice about the shape of the two curves? What happens to the number of operations when $n$ is multiplied by 10?
>
> *Your answer:* both the curve of the # of operations, and curve of running time are lines. when n grows 10 times, the # of operations grows 10 times as well since it's proportional to the size of the input

### Exercise 1 — Average
Write `average(T)` that returns the average of a non-empty list, **without** using `sum()`.

````python
def average(T):
    # TO DO
    sum = 0
    for n in T:
        sum += n
    return sum/len(T)
    pass

check("average([2, 4, 6])", average([2, 4, 6]), 4)
check("average([5])", average([5]), 5)
````

### Exercise 2 — Count occurrences
Write `occurrences(T, x)` that returns how many times `x` appears in `T`, **without** using `T.count()`.

````python
def occurrences(T, x):
    # TO DO
    count = 0
    for i in range(len(T)):
        if x == T[i]:
            count += 1
    return count
    pass

check("occurrences([1,2,1,3,1], 1)", occurrences([1, 2, 1, 3, 1], 1), 3)
check("occurrences([], 7)", occurrences([], 7), 0)
````

### Exercise 3 — Reverse a list
Write `reverse(T)` that returns a **new** reversed list, **without** `reversed()` or `T[::-1]`.

````python
def reverse(T):
    # TO DO
    n = len(T)
    for i in range(n // 2):
        temp = T[i]
        T[i] = T[n - 1 - i]
        T[n - 1 - i] = temp
    return T
    pass

check("reverse([1, 2, 3])", reverse([1, 2, 3]), [3, 2, 1])
check("reverse([])", reverse([]), [])
````

### Exercise 4 — Letter frequencies with a dictionary
Write `frequencies(word)` that returns a dictionary `{letter: number of appearances}`.

````python
def frequencies(word):
    # TO DO
    freq = {}
    for letter in word:
        if letter not in freq:
            freq[letter] = 1
        else:
            freq[letter] += 1
    return freq
    pass

check("frequencies('banana')", frequencies("banana"), {"b": 1, "a": 3, "n": 2})
````

### Exercise 5 — Recursive power
Write a **recursive** `power(a, n)` that computes $a^n$ for $n \ge 0$, without `**` or `pow()`.

````python
def power(a, n):
    # TO DO
    if n == 0:
        return 1
    return a * power(a, n-1)
    pass

check("power(2, 10)", power(2, 10), 1024)
check("power(5, 0)", power(5, 0), 1)
````

### Exercise 6 — Measure and compare ⭐
Here are two functions that test whether a list contains a **duplicate**.

1. Add a **comparison counter** to each one (the elementary operation is the `==` or `in` comparison).
2. Run them for `n = 10, 100, 1000, 2000` on a list **with no duplicates** (`list(range(n))`).
3. Plot the **number of comparisons** of both functions on the same **log-log** chart.
4. Describe **in your own words** (without $O(\cdot)$ notation) what you observe when $n$ is multiplied by 10.

````python
def has_duplicate_v1(T):
    counter = 0
    for i in range(len(T)):
        for j in range(i + 1, len(T)):
            # TO DO: increment the counter
            counter += 1
            if T[i] == T[j]:
                return True, counter
    return False, counter

def has_duplicate_v2(T):
    counter = 0
    seen = set()
    for e in T:
        # TO DO: increment the counter
        counter += 1
        if e in seen:
            return True, counter
        seen.add(e)
    return False, counter
````

````python
# TO DO: loop over the sizes, collect the counters, log-log plot
lengths, v1_comparisons , v2_comparisons = [], [], []
for n in [10, 100, 1_000, 2_000]:
    vect = list(range(n))
    _ , counter = has_duplicate_v1(vect)
    _ , counter2 = has_duplicate_v2(vect)
    v1_comparisons.append(counter)
    v2_comparisons.append(counter2)
    lengths.append(n)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4))

ax1.loglog(lengths, v1_comparisons, "o-")
ax1.set(title="Number of comparisons for has_duplicate_v1", xlabel="n (input size)", ylabel="comparisons")
ax1.grid(True, which="both", alpha=0.3)

ax2.loglog(lengths, v2_comparisons, "o-", color="tab:red")
ax2.set(title="Number of comparisons for has_duplicate_v2", xlabel="n (input size)", ylabel="comparisons")
ax2.grid(True, which="both", alpha=0.3)

plt.tight_layout()
plt.show()
print(v1_comparisons, v2_comparisons)
````

> *Your observations:* dup1 grows is growing quadratically ($ n(n - 1)/2 $ to be exact), whereas dup2 grows linearly
